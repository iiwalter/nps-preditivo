#### Entendimento do negócio

#### Qual problema de negócio está sendo resolvido?

O problema central que este projeto resolve é a dependência exclusiva da pesquisa de NPS como mecanismo de escuta do cliente. Essa dependência cria um ciclo tardio: a empresa só toma conhecimento da insatisfação quando o cliente já decidiu se é promotor ou detrator. O NPS Preditivo quebra esse ciclo ao antecipar o risco de insatisfação com base em sinais operacionais, permitindo que as áreas responsáveis ajam antes do veredicto do cliente, transformando potenciais detratores em clientes fidelizados.

#### Por que o NPS é importante para um e-commerce?

Porque é a única métrica que captura essa percepção de forma consolidada. Uma nota baixa não aponta apenas um problema isolado, ela reflete o peso acumulado de toda a operação sobre a experiência do cliente. Com isso, o NPS permite localizar se a operação está gerando detratores ou promotores, e em qual parte da jornada o problema se concentra. Para um e-commerce que compete por recompra e indicação, monitorar essa métrica de forma contínua é condição básica para crescer de forma sustentável

#### Quais áreas poderiam se beneficiar desses insights?
Logística
* A logística é uma das áreas com maior volume de variáveis capazes de gerar insatisfação: prazo de entrega, atrasos e valor do frete impactam diretamente a experiência do cliente. Com o modelo preditivo, é possível identificar quais transportadoras, rotas ou regiões concentram maior risco de gerar detratores, permitindo ajustes operacionais antes que o problema se repita em escala. (veremos isso ao decorrer do EDA)

Atendimento ao Cliente
* O atendimento é a principal linha de reversão de uma experiência negativa. Com o modelo preditivo, a área deixa de agir apenas quando o cliente reclama e passa a antecipar o contato com clientes de alto risco, identificados por rota, transportadora ou histórico de ocorrências. Essa abordagem proativa transforma uma experiência que seria negativa em uma oportunidade de fidelização.

Marketing e Gestão
* A satisfação do cliente impacta diretamente a percepção da marca e as decisões de crescimento do negócio. Com os insights do modelo preditivo, marketing e gestão conseguem entender quais fatores mais influenciam a experiência, orientando campanhas, prioridades de investimento e estratégias de retenção com base em dados reais da operação.

Como o NPS impacta:
1. Recompra:
Clientes promotores tendem a comprar novamente sem esforço adicional de aquisição, enquanto detratores dificilmente retornam e são mais caros de recuperar.

2. Boca a boca:
Promotores geram recomendações espontâneas (aquisição orgânica), enquanto detratores espalham experiências negativas que impactam a marca de forma difícil de controlar.

3. Market share:
Um NPS alto fortalece retenção, indicação e percepção de valor, resultando em ganho de participação de mercado no longo prazo.

Quais indicadores de mercado poderiam complementar essa análise?

1. Benchmarks de NPS:
Comparam seu NPS com o mercado para contextualizar o desempenho e identificar gaps competitivos.

2. Taxa de recompra e LTV:
Validam se a intenção medida pelo NPS se traduz em comportamento real e geração de receita ao longo do tempo.

3. CSAT e CES:
Complementam o NPS ao detalhar a experiência: o CSAT mede satisfação em pontos específicos e o CES avalia o esforço do cliente, tornando a análise mais acionável.

### Definição da Target: qual é o alvo desse problema de negócio?

A variável que representa a satisfação do cliente é a **Nota do NPS**

Ela foi escolhida porque representa a satisfação do cliente de forma consolidada toda a jornada que o cliente teve durante a compra em nosso e-commerce. Impacta diretamente na futura decisão do cliente de realizar novas compras e indicar a marca para outras pessoas.

Ela é coletada no final da jornada de compra, após o cliente ter passado por toda a experiência na plataforma, como adição de produtos, pagamentos, etc.. Isso porquê o NPS tem como objetivo captar a percepção final sobre a experiência completa.

O maior risco é usar a nota como uma verdade absoluta isoladamente, levando em consideração que a maioria das pessoas que respondem as pesquisas de NPS são muito satisfeitas ou muito insatisfeitas, fica uma lacuna de dados em relação a experiência média que os usuários tem na plataforma. Usando de forma inadequada a empresa pode trabalhar apenas pensando nesses casos onde estão nas extremidades, sem levar em consideração a experiência média dos usuários.

#### Construindo uma amostra aleatória para estimar a média de NPS com incerteza controlada.

```python
np.random.seed(42)
nps_data = nps_data.sample(n=1000, replace=False)
sns.histplot(nps_data['nps_score'], kde=True)
```

Aqui, obtemos um resumo rápido da satisfação geral dos nossos clientes (o 'nps_score'), incluindo a média, o valor mais alto e o mais baixo. É um raio-x da percepção dos clientes.

```python
display(nps_data['nps_score'].describe())
```

Este gráfico nos mostra visualmente se atrasos na entrega (`delivery_delay_days`) afetam a satisfação do cliente (`nps_score`). Queremos entender se há uma relação direta entre o tempo de espera e a percepção do serviço.

```python
fig = px.scatter(nps_data, x='delivery_delay_days', y='nps_score', title='Correlação entre Atraso na Entrega e NPS Score')
fig.show()
```

Criamos uma 'regra' para categorizar a satisfação dos clientes com base no `nps_score` deles. Assim, podemos classificá-los em grupos como 'Promotores', 'Passivo' e 'Detratores'.

```python
def nps_score_class(df):
    if df['nps_score'] >= 9:
        return 'Promotores'
    elif df['nps_score'] >= 7 <= 8:
        return 'Passivos'
    else:
        return 'Detratores'
```

Aplicamos essa 'regra' a todos os nossos clientes, criando uma nova coluna que mostra a classificação de satisfação de cada um. Isso nos permite ver, por exemplo, quantos clientes são 'Detratores'.

```python
nps_data['nps_class'] = nps_data.apply(nps_score_class, axis=1)
nps_data
```

Contamos quantos clientes caem em cada categoria de satisfação. Isso nos dá uma visão clara da distribuição dos níveis de satisfação.

```python
nps_data['nps_class'].value_counts()
```

Calculamos a média do tempo de entrega, atrasos e satisfação (NPS) para cada região. Essa informação é crucial para identificar áreas com melhor ou pior desempenho.

```python
nps_data_region_delay = pd.DataFrame(nps_data.groupby('customer_region')[['delivery_time_days','delivery_delay_days','nps_score']].mean().round(2)).reset_index()
```

Mostramos a tabela com as médias de tempo de entrega, atrasos e NPS para cada região, facilitando a comparação entre elas.

```python
display(nps_data_region_delay)
```

Aqui, vemos quantos clientes de cada categoria de satisfação existem em cada região. Isso ajuda a entender onde temos mais problemas ou sucessos.

```python
nps_data_region_class_counts = nps_data.groupby('customer_region')['nps_class'].value_counts().reset_index(name='count')
display(nps_data_region_class_counts)
```

Este gráfico, chamado 'mapa de calor', nos mostra visualmente as relações entre os fatores e o NPS. Cores mais fortes indicam uma ligação mais intensa, positiva ou negativa.

```python
influence_columns = [
    'complaints_count',
    'resolution_time_days',
    'customer_service_contacts',
    'delivery_delay_days',
    'delivery_time_days',
    'nps_score'
]

influence_correlation_matrix = nps_data[influence_columns].corr()
display(influence_correlation_matrix)
```

Focamos nos clientes 'Detratores'. Calculamos a média dos problemas (atrasos, reclamações, etc.) *apenas* para eles, para entender o que mais os incomoda e gera essa insatisfação extrema.

```python
plt.figure(figsize=(10, 8))
sns.heatmap(influence_correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Matriz de Correlação entre Fatores de Atendimento/Entrega e NPS Score')
plt.show()
```

## Quais fatores parecem mais críticos para a satisfação?

1. Atraso na entrega: Com uma correlação de **-0.597**, é o fator mais crítico. Grandes atrasos na entrega têm um impacto muito significativo na insatisfação do cliente.
2. Número de reclamações: Com uma correlação de **-0.497**, é o segundo fator mais impactante. Um número elevado de reclamações está fortemente associado a um NPS mais baixo.
3. Contatos com o serviço de atendimento ao cliente: Com uma correlação de **-0.351**, sugere que quanto mais o cliente precisa contatar o serviço, menor tende a ser sua satisfação.
4. Tempo de resolução de problemas: Embora menos impactante que os anteriores com **-0.191**, ainda é um fator negativo. Quanto maior o tempo para resolver um problema, menor a satisfação.






Média dos fatores influenciadores para clientes 'Detratores'

```python
detractors_data = nps_data[nps_data['nps_class'] == 'Detratores']

detractor_influencing_factors_mean = detractors_data[influence_columns].mean().drop('nps_score')
display(detractor_influencing_factors_mean)
```

Média geral dos fatores influenciadores

```python
overall_influencing_factors_mean = nps_data[influence_columns].mean().drop('nps_score')
display(overall_influencing_factors_mean)
```

Este gráfico compara visualmente a média dos problemas para clientes 'Detratores' versus a média geral. Ele mostra de forma clara quais fatores estão desproporcionalmente piores para os insatisfeitos, indicando onde devemos agir.

```python

comparison_df = pd.DataFrame({
    'Fator': detractor_influencing_factors_mean.index,
    'Média Detratores': detractor_influencing_factors_mean.values,
    'Média Geral': overall_influencing_factors_mean.values
})

comparison_df_melted = comparison_df.melt(id_vars='Fator', var_name='Grupo', value_name='Média dos Fatores')

plt.figure(figsize=(12, 7))
sns.barplot(x='Fator', y='Média dos Fatores', hue='Grupo', data=comparison_df_melted, palette='viridis')
plt.title('Comparação da Média dos Fatores Influenciadores: Detratores vs. Geral')
plt.ylabel('Média')
plt.xlabel('Fator Influenciador')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Grupo')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## O que mais gera detratores?

* Atraso na Entrega: A média de dias de atraso na entrega para detratores é de 2.44 dias, significativamente maior do que a média geral de 2.19 dias. Isso indica que o atraso na entrega é o maior impulsionador para detratores.

* Número de reclamações: A média de reclamações para os clientes detratores é de 4.45, que é maior do que a média geral de 4.15. Isso indica que um maior volume de reclamações está associado a uma maior insatisfação do cliente


Aqui, analisamos apenas os clientes que *não* compraram novamente nos últimos 30 dias. Calculamos como os problemas (atrasos, reclamações, etc.) se relacionam com a satisfação *neste grupo específico*. O objetivo é entender o que os impede de voltar a comprar.

Qual fator mais critico para clientes que não fizeram recompra?

Análise de Clientes com e sem Recompra + NPS

```python
repeat_purchase_data = nps_data[nps_data['repeat_purchase_30d'] >= 1]
no_repeat_purchase_data = nps_data[nps_data['repeat_purchase_30d'] == 0]

promoters_repeat = repeat_purchase_data[repeat_purchase_data['nps_class'] == 'Promotores']
promoters_no_repeat = no_repeat_purchase_data[no_repeat_purchase_data['nps_class'] == 'Promotores']
detractors_no_repeat = no_repeat_purchase_data[no_repeat_purchase_data['nps_class'] == 'Detratores']
detractors_repeat = repeat_purchase_data[repeat_purchase_data['nps_class'] == 'Detratores']
passives_repeat = repeat_purchase_data[repeat_purchase_data['nps_class'] == 'Passivos']
passives_no_repeat = no_repeat_purchase_data[no_repeat_purchase_data['nps_class'] == 'Passivos']

display(f"Número de Promotores sem recompra: {len(promoters_no_repeat)}")
display(f"Número de Promotores com recompra: {len(promoters_repeat)}")
display(f"Número de Detratores sem recompra: {len(detractors_no_repeat)}")
display(f"Número de Detratores com recompra: {len(detractors_repeat)}")
display(f"Número de Passivos sem recompra: {len(passives_no_repeat)}")
display(f"Número de Passivos com recompra: {len(passives_repeat)}")
```

Podemos analisar que:
* Promotores sempre recompram
* Passivos têm aproximadamente 42% de probabilidade de recompra
* Detratores nunca realizam uma recompra.


Comparação das Médias de Características Relevantes para Recompra

Vamos analisar as médias das mesmas características chaves para os grupos que destacaram-se na quantidade de recompras:
*   Promotores (sempre fizeram uma recompra)
*   Detratores (nunca fizeram uma recompra)



```python
comparison_columns = [
    'customer_age',
    'customer_tenure_months',
    'items_quantity',
    'discount_value',
    'payment_installments',
    'freight_value',
    'customer_service_contacts',
    'resolution_time_days',
    'complaints_count',
    'csat_internal_score',
    'delivery_delay_days'
]

repeat_means = promoters_repeat[comparison_columns].mean().rename('Promoters_Repeat_Mean')
no_repeat_means = detractors_no_repeat[comparison_columns].mean().rename('Detractors_No_Repeat_Mean')

comparison_df_nps_repeat = pd.DataFrame({
    'Característica': repeat_means.index,
    'Média Com Recompra': repeat_means.values,
    'Média Sem Recompra': no_repeat_means.values,
})

display(comparison_df_nps_repeat.round(2))
```

Visualização da Comparação de Recompra

Um gráfico de barras para visualizar as diferenças entre os grupos:
* Promotores (sempre fizeram uma recompra)
* Detratores (nunca fizeram uma recompra)

```python
comparison_df_melted_nps_repeat = comparison_df_nps_repeat.melt(id_vars='Característica', var_name='Grupo', value_name='Média')

plt.figure(figsize=(15, 8))
sns.barplot(x='Característica', y='Média', hue='Grupo', data=comparison_df_melted_nps_repeat, palette='viridis')
plt.title('Comparação de Características: Promotores vs. Detratores')
plt.xlabel('Característica')
plt.ylabel('Média')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Grupo')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

Análise de Distribuição Geográfica focada em Recompra

Vamos verificar a distribuição das regiões para
* Promotores (sempre fizeram uma recompra)
* Detratores (nunca fizeram uma recompra)

```python
promoters_region_dist = promoters_repeat['customer_region'].value_counts(normalize=True).mul(100).round(2).to_frame(name='% Recompra').rename_axis('Região').reset_index()
detractors_region_dist = detractors_no_repeat['customer_region'].value_counts(normalize=True).mul(100).round(2).to_frame(name='% Sem Recompra').rename_axis('Região').reset_index()

combined_region_dist = pd.merge(promoters_region_dist, detractors_region_dist, on='Região', how='outer').fillna(0)

display(combined_region_dist)
```
